# Exploration et Visualisation des Données COPD

```python
# Statistiques sur la variable cible
print("📊 Statistiques de la variable 'Value':")
print(f"{'='*70}")
print(f"Moyenne: {df['Value'].mean():.4f}")
print(f"Médiane: {df['Value'].median():.4f}")
print(f"Écart-type: {df['Value'].std():.4f}")
print(f"Min: {df['Value'].min():.4f}")
print(f"Max: {df['Value'].max():.4f}")
print(f"Quartile 25%: {df['Value'].quantile(0.25):.4f}")
print(f"Quartile 75%: {df['Value'].quantile(0.75):.4f}")

# Distribution
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Histogramme
axes[0].hist(df['Value'].dropna(), bins=50, edgecolor='black', alpha=0.7, color='steelblue')
axes[0].axvline(df['Value'].mean(), color='red', linestyle='--', linewidth=2, label=f'Moyenne: {df["Value"].mean():.2f}')
axes[0].axvline(df['Value'].median(), color='green', linestyle='--', linewidth=2, label=f'Médiane: {df["Value"].median():.2f}')
axes[0].set_xlabel('Value', fontsize=11)
axes[0].set_ylabel('Fréquence', fontsize=11)
axes[0].set_title('Distribution de Value (Histogramme)', fontsize=12, fontweight='bold')
axes[0].legend()
axes[0].grid(alpha=0.3)

# Box plot
axes[1].boxplot(df['Value'].dropna(), vert=True)
axes[1].set_ylabel('Value', fontsize=11)
axes[1].set_title('Distribution de Value (Box Plot)', fontsize=12, fontweight='bold')
axes[1].grid(alpha=0.3)

# Violin plot
axes[2].violinplot([df['Value'].dropna()], vert=True, showmeans=True, showmedians=True)
axes[2].set_ylabel('Value', fontsize=11)
axes[2].set_title('Distribution de Value (Violin Plot)', fontsize=12, fontweight='bold')
axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

## 6. Évolution temporelle

```python
# Évolution temporelle globale (moyenne par année)
yearly_stats = df.groupby('Year')['Value'].agg(['mean', 'median', 'std', 'count']).reset_index()

fig, axes = plt.subplots(2, 1, figsize=(14, 10))

# Graphique 1: Moyenne et médiane
axes[0].plot(yearly_stats['Year'], yearly_stats['mean'], marker='o', linewidth=2.5, 
             markersize=7, label='Moyenne', color='steelblue')
axes[0].plot(yearly_stats['Year'], yearly_stats['median'], marker='s', linewidth=2.5, 
             markersize=7, label='Médiane', color='coral')
axes[0].fill_between(yearly_stats['Year'], 
                      yearly_stats['mean'] - yearly_stats['std'],
                      yearly_stats['mean'] + yearly_stats['std'],
                      alpha=0.2, color='steelblue', label='±1 écart-type')
axes[0].set_xlabel('Année', fontsize=12)
axes[0].set_ylabel('Value', fontsize=12)
axes[0].set_title('Évolution temporelle globale de COPD (Moyenne ± Écart-type)', fontsize=14, fontweight='bold')
axes[0].legend(fontsize=10)
axes[0].grid(alpha=0.3)

# Graphique 2: Nombre d'observations
axes[1].bar(yearly_stats['Year'], yearly_stats['count'], color='lightgreen', edgecolor='black', alpha=0.7)
axes[1].set_xlabel('Année', fontsize=12)
axes[1].set_ylabel('Nombre d\'observations', fontsize=12)
axes[1].set_title('Nombre d\'observations par année', fontsize=14, fontweight='bold')
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

# Analyse de la tendance
trend_change = yearly_stats['mean'].iloc[-1] - yearly_stats['mean'].iloc[0]
trend_pct = (trend_change / yearly_stats['mean'].iloc[0]) * 100

print(f"📈 Analyse de la tendance:")
print(f"{'='*70}")
print(f"Valeur initiale ({yearly_stats['Year'].iloc[0]}): {yearly_stats['mean'].iloc[0]:.4f}")
print(f"Valeur finale ({yearly_stats['Year'].iloc[-1]}): {yearly_stats['mean'].iloc[-1]:.4f}")
print(f"Changement: {trend_change:.4f} ({trend_pct:+.2f}%)")
if trend_change > 0:
    print("➡️ Tendance: AUGMENTATION")
else:
    print("➡️ Tendance: DIMINUTION")
```

## 9. Analyse des corrélations (Features numériques)

```python
# Identifier les colonnes numériques
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

# Retirer les colonnes d'identification
id_cols = ['Year']
feature_cols = [col for col in numeric_cols if col not in id_cols]

print(f"📊 Colonnes numériques à analyser: {len(feature_cols)}")

# Calculer les corrélations avec Value
if 'Value' in feature_cols:
    correlations = df[feature_cols].corr()['Value'].drop('Value').abs().sort_values(ascending=False)
    
    # Top 15 corrélations
    top_15_corr = correlations.head(15)
    
    print(f"\n🔗 Top 15 features corrélées avec 'Value':")
    print(f"{'='*70}")
    for feat, corr in top_15_corr.items():
        print(f"{feat[:50]:50s} : {corr:.4f}")
    
    # Visualisation
    plt.figure(figsize=(12, 8))
    plt.barh(range(len(top_15_corr)), top_15_corr.values, color='teal', edgecolor='black')
    plt.yticks(range(len(top_15_corr)), [f[:40] for f in top_15_corr.index], fontsize=10)
    plt.xlabel('Corrélation absolue avec Value', fontsize=12)
    plt.title('Top 15 Features - Corrélation avec Value', fontsize=14, fontweight='bold')
    plt.grid(axis='x', alpha=0.3)
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.show()
else:
    print("⚠️ Colonne 'Value' non trouvée dans les données numériques")
```

```python
# Matrice de corrélation pour top features
if 'Value' in feature_cols and len(top_15_corr) > 0:
    # Sélectionner top features + Value
    selected_features = list(top_15_corr.head(10).index) + ['Value']
    
    # Calculer matrice de corrélation
    corr_matrix = df[selected_features].corr()
    
    # Heatmap
    plt.figure(figsize=(12, 10))
    sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0,
                square=True, linewidths=0.5, cbar_kws={"shrink": 0.8},
                xticklabels=[f[:30] for f in corr_matrix.columns],
                yticklabels=[f[:30] for f in corr_matrix.index])
    plt.title('Matrice de corrélation - Top 10 Features vs Value', fontsize=14, fontweight='bold')
    plt.xticks(rotation=45, ha='right', fontsize=9)
    plt.yticks(fontsize=9)
    plt.tight_layout()
    plt.show()
```

```python
seuil = 0.8

# Afficher les couples de variables avec corrélation absolue > seuil
corr = df[numeric_cols].corr()  # utilise numeric_cols défini précédemment
corr_abs = corr.abs()

# Masque pour ne garder que la triangle supérieure (éviter doublons et diagonale)
mask = np.triu(np.ones(corr.shape), k=1).astype(bool)
high = corr_abs.where(mask)

# Construire la liste des paires et trier
high_pairs = high.stack().reset_index()
high_pairs.columns = ['var1', 'var2', 'abs_corr']
high_pairs = high_pairs[high_pairs['abs_corr'] > seuil].sort_values('abs_corr', ascending=False)

if high_pairs.empty:
    print(f"Aucun couple de variables avec corrélation absolue > {seuil} trouvé.")
else:
    print(f"Couples de variables avec corrélation absolue > {seuil} :\n")
    for _, row in high_pairs.iterrows():
        v1, v2, a = row['var1'], row['var2'], row['abs_corr']
        actual = corr.loc[v1, v2]
        print(f"{v1}  <->  {v2} : corr = {actual:.4f} (abs = {a:.4f})")
```

## 11. Scatter plots (Value vs Top Features)

```python
# Scatter plots pour les 6 features les plus corrélées
if 'Value' in feature_cols and len(top_15_corr) > 0:
    top_6_features = top_15_corr.head(6).index
    
    fig, axes = plt.subplots(2, 3, figsize=(18, 10))
    axes = axes.ravel()
    
    for idx, feature in enumerate(top_6_features):
        # Scatter plot
        axes[idx].scatter(df[feature], df['Value'], alpha=0.5, s=30, color='steelblue', edgecolors='black', linewidth=0.5)
        
        # Ligne de régression
        mask = df[[feature, 'Value']].notna().all(axis=1)
        if mask.sum() > 1:
            z = np.polyfit(df.loc[mask, feature], df.loc[mask, 'Value'], 1)
            p = np.poly1d(z)
            x_line = np.linspace(df[feature].min(), df[feature].max(), 100)
            axes[idx].plot(x_line, p(x_line), "r--", linewidth=2, label='Régression linéaire')
        
        axes[idx].set_xlabel(feature[:40], fontsize=10)
        axes[idx].set_ylabel('Value', fontsize=10)
        axes[idx].set_title(f'Value vs {feature[:30]}\n(corr: {correlations[feature]:.3f})', 
                           fontsize=11, fontweight='bold')
        axes[idx].grid(alpha=0.3)
        axes[idx].legend(fontsize=8)
    
    plt.tight_layout()
    plt.show()
else:
    print("⚠️ Pas assez de features pour générer les scatter plots")
```
